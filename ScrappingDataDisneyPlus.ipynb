## Import Library

```python
import pandas as pd
from google_play_scraper import reviews, Sort
```

## Scrapping data

```python
raw_reviews,continuation_token = reviews(
    'in.startv.hotstar.dplus',
    lang='id',
    country='id',
    sort=Sort.MOST_RELEVANT,
    count=30000
)

len(raw_reviews)
```

## Ubah Hasil Scrapping menjadi DataFrame

```python
df = pd.DataFrame(raw_reviews)
df.head()
```

## Seleksi Fitur
Fitur yang akan digunakan:
- reviewId (untuk pengecekan duplikat)
- score / rating
- content / ulasan

```python
df = df[["reviewId","score", "content"]]

df.head()
```

## Cek Duplikat dan Data Null

```python
print("Jumlah data:", len(df))
print("\nJumlah duplikat reviewId:", df.duplicated(subset=["reviewId"]).sum())
print("\nMissing values:")
print(df.isnull().sum())
```

## Cek Distribusi Rating

```python
df["score"].value_counts().sort_index()
```

## Simpan menjadi file CSV

```python
OUTPUT_CSV = "disneyplus_reviews_raw.csv"
df.to_csv(OUTPUT_CSV, index=False, encoding="utf-8-sig")
print(f"File berhasil disimpan: {OUTPUT_CSV}")
```
